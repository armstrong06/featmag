```python
import pandas as pd
import numpy as np
```

```python
s_gains = pd.read_csv("../files/db_featmag_gains_S.csv")
p_gains = pd.read_csv("../files/db_featmag_gains_P.csv")
```

```python
s_gains["phase"] = "S"
p_gains["phase"] = "P"
```

```python
merged_df = s_gains[["channel_id", "featmag_gain"]].merge(p_gains[["channel_id", "featmag_gain"]], on=["channel_id"], suffixes=("S", "P"))
# Use the P row if there is a value in the P catalog but not the S - should just be 1C stations or some that have P but no S picks
p_ids_to_use = merged_df[(~np.isnan(merged_df["featmag_gainP"])) & (np.isnan(merged_df["featmag_gainS"]))]["channel_id"]
merged_summ_df = pd.concat([s_gains[~s_gains["channel_id"].isin(p_ids_to_use)], p_gains[p_gains["channel_id"].isin(p_ids_to_use)]]).sort_values("channel_id")
assert len(merged_summ_df) == len(s_gains)
```
